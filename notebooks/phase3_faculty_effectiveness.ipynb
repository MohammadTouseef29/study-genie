```python
import pandas as pd
import numpy as np

np.random.seed(42)
```

```python
activity_df = pd.read_csv(
    "../data/analytics/student_activity.csv",
    parse_dates=["date"]
)

quiz_df = pd.read_csv(
    "../data/analytics/quiz_performance.csv",
    parse_dates=["date"]
)

activity_df.head()
```

```python
lecture_metrics = (
    activity_df
    .groupby("lecture_id")
    .agg(
        avg_engagement=("engagement_score", "mean"),
        attendance_rate=("attendance", "mean")
    )
    .reset_index()
)

lecture_metrics.head()
```

```python
lecture_metrics["eng_norm"] = (
    lecture_metrics["avg_engagement"] - lecture_metrics["avg_engagement"].min()
) / (lecture_metrics["avg_engagement"].max() - lecture_metrics["avg_engagement"].min())

lecture_metrics["att_norm"] = (
    lecture_metrics["attendance_rate"] - lecture_metrics["attendance_rate"].min()
) / (lecture_metrics["attendance_rate"].max() - lecture_metrics["attendance_rate"].min())

lecture_metrics.head()
```

```python
lecture_metrics["learning_impact"] = np.random.uniform(
    0.6, 0.95, size=len(lecture_metrics)
)

lecture_metrics.head()
```

```python
lecture_metrics["effectiveness_score"] = (
    lecture_metrics["eng_norm"]
    + lecture_metrics["att_norm"]
    + lecture_metrics["learning_impact"]
) / 3

lecture_metrics.sort_values(
    "effectiveness_score", ascending=False
).head()
```

```python
lecture_metrics.to_csv(
    "../data/analytics/faculty_effectiveness.csv",
    index=False
)

print("✅ Faculty effectiveness analytics saved")
```
